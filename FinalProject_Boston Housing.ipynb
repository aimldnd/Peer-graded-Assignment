```python
import pandas as pd
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as pyplot
import scipy.stats
import statsmodels.api as sm
from statsmodels.formula.api import ols
```

#### Read the dataset in the csv file from the URL


```python
from js import fetch
import io

URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ST0151EN-SkillsNetwork/labs/boston_housing.csv'
resp = await fetch(URL)
boston_url = io.BytesIO((await resp.arrayBuffer()).to_py())
```

```python
boston_df=pd.read_csv(boston_url)
```

#### Add your code below following the instructions given in the course to complete the peer graded assignment


```python
# CHARLES RIVER BOUNDARY VS MEDIAN VALUE
pyplot.figure(figsize=(8, 5))
sns.boxplot(x='CHAS', y='MEDV', data=boston_df, palette='Set2')
pyplot.title('Boxplot of Median Value vs Charles River Boundary')
pyplot.xlabel('Bounded by Charles River (CHAS: 0 = No, 1 = Yes)')
pyplot.ylabel('Median Value ($1,000s) [MEDV]')
pyplot.show()

chas_0 = boston_df[boston_df['CHAS'] == 0]['MEDV']
chas_1 = boston_df[boston_df['CHAS'] == 1]['MEDV']

t_stat, p_val = scipy.stats.ttest_ind(chas_1, chas_0, equal_var=False)
print(f"T-statistic: {t_stat:.4f}, P-value: {p_val:.4f}")
print(f"CHAS = 0 Median: {chas_0.median():.2f}")
print(f"CHAS = 1 Median: {chas_1.median():.2f}")
```

```python
# Proportion of Older units vs Median Value

boston_df.loc[(boston_df['AGE'] <= 35), 'age_group'] = '35 years and younger'
boston_df.loc[(boston_df['AGE'] > 35) & (boston_df['AGE'] < 70), 'age_group'] = 'between 35 and 70 years'
boston_df.loc[(boston_df['AGE'] >= 70), 'age_group'] = '70 years and older'

pyplot.figure(figsize=(9, 5))
sns.boxplot(x='age_group', y='MEDV', data=boston_df, order=['35 years and younger', 'between 35 and 70 years', '70 years and older'], palette='Set3')
pyplot.title('Median Value vs Proportion of Older Units (AGE)')
pyplot.xlabel('Age Group')
pyplot.ylabel('Median Value ($1,000s) [MEDV]')
pyplot.show()
```

```python
# Nitric Oxide Concentration vs. Non-Retail Business Acres
pyplot.figure(figsize=(8, 5))
sns.scatterplot(x='INDUS', y='NOX', data=boston_df, alpha=0.6)
sns.regplot(x='INDUS', y='NOX', data=boston_df, scatter=False, color='green')
pyplot.title('Nitric Oxide Concentration vs. Non-Retail Business Acres')
pyplot.xlabel('Proportion of Non-Retail Business Acres per Town (INDUS)')
pyplot.ylabel('Nitric Oxide Concentration (parts per 10 million) [NOX]')
pyplot.show()

corr, p_val = scipy.stats.pearsonr(boston_df['INDUS'], boston_df['NOX'])
print(f"Pearson Correlation: {corr:.4f}, P-value: {p_val:.4e}")
```

```python
# Weighted Distance vs. Median Value
pyplot.figure(figsize=(8, 5))
sns.histplot(boston_df['DIS'], bins=30, kde=True, color='purple')
pyplot.title('Histogram of Weighted Distances to Employment Centers')
pyplot.xlabel('Weighted Distance (DIS)')
pyplot.ylabel('Frequency')
pyplot.show()
```
